```python
# text to numeric

data['TotalCharges'] = pd.to_numeric(data['TotalCharges'], errors='coerce')
data['TotalCharges'] = data['TotalCharges'].fillna(0)
```

```python
# Numeric feature vs Churn
for col in ['tenure', 'MonthlyCharges', 'TotalCharges']:
    plt.figure(figsize=(7,3))
    sns.kdeplot(data=data, x=col, hue='Churn', fill=True, common_norm=False)
    plt.title(f'{col} Distribution by Churn')
    plt.show()


```

```python
# feature engineering
# Encode target
#Feature Transformation 
data['Churn'] = data['Churn'].map({'Yes': 1, 'No': 0})

# One-hot encode categoricals
X = pd.get_dummies(data.drop('Churn', axis=1), drop_first=True)

# Scale numeric (for Logistic Regression)
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X[['tenure','MonthlyCharges','TotalCharges']] = scaler.fit_transform(X[['tenure','MonthlyCharges','TotalCharges']])
```

```python
data['Churn'].isnull().sum()
data['Churn'].unique()
```

```python
# 2. Feature Construction
# Tenure groups (binning)
data['tenure_group'] = pd.cut(data['tenure'], bins=[0,12,24,48,72], 
                              labels=['0-1yr','1-2yr','2-4yr','4-6yr'])

# Total services subscribed (count of "Yes" across service columns)
service_cols = ['PhoneService','OnlineSecurity','OnlineBackup','DeviceProtection',
                'TechSupport','StreamingTV','StreamingMovies']
data['total_services'] = (data[service_cols] == 'Yes').sum(axis=1)

# Average monthly spend ratio
data['charge_per_tenure'] = data['TotalCharges'] / (data['tenure'] + 1)  # +1 avoids div by 0

# Household type interaction
data['household_type'] = data['Partner'] + '_' + data['Dependents']



```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, stratify=y, random_state=42)
```

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(max_iter=5000)
model.fit(X_train, y_train)
```

```python
from sklearn.metrics import confusion_matrix, classification_report

y_pred = model.predict(X_test)

print(confusion_matrix(y_test, y_pred))
print(classification_report(y_test, y_pred))
```

```python
from xgboost import XGBClassifier
from sklearn.metrics import roc_auc_score

scale_pos_weight = (y_train==0).sum() / (y_train==1).sum()
print("scale_pos_weight:", scale_pos_weight)

xgb_model = XGBClassifier(scale_pos_weight=scale_pos_weight, eval_metric='auc', random_state=42)
xgb_model.fit(X_train, y_train)

y_pred_xgb = xgb_model.predict(X_test)
y_proba_xgb = xgb_model.predict_proba(X_test)[:,1]

print(confusion_matrix(y_test, y_pred_xgb))
print(classification_report(y_test, y_pred_xgb))
print("ROC-AUC:", roc_auc_score(y_test, y_proba_xgb))
```

```python
from sklearn.model_selection import RandomizedSearchCV

param_dist = {
    'n_estimators': [100, 200, 300],
    'max_depth': [3, 4, 5, 6],
    'learning_rate': [0.01, 0.05, 0.1, 0.2],
    'subsample': [0.7, 0.8, 1.0],
    'colsample_bytree': [0.7, 0.8, 1.0]
}

xgb_base = XGBClassifier(scale_pos_weight=scale_pos_weight, eval_metric='auc', random_state=42)

search = RandomizedSearchCV(xgb_base, param_dist, n_iter=20, scoring='roc_auc', cv=5, random_state=42, n_jobs=-1)
search.fit(X_train, y_train)

print("Best params:", search.best_params_)
best_model = search.best_estimator_

y_pred_best = best_model.predict(X_test)
y_proba_best = best_model.predict_proba(X_test)[:,1]
print(classification_report(y_test, y_pred_best))
print("ROC-AUC:", roc_auc_score(y_test, y_proba_best))
```
